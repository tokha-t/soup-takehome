## Pinned data, prompts and deterministic scorers

```python
%%writefile /kaggle/temp/soup_bakeoff/probe_data.py
"""Pinned candidate data, formatting, scoring and token-length checks."""

from dataclasses import dataclass, field
import json
import math
import random
import re


SEED = 42
MODEL_ID = "Qwen/Qwen2.5-1.5B-Instruct"
MODEL_REVISION = "989aa7980e4cf806f80c7fef2b1adb7bc71aa306"
TRAIN_ROWS = 500
HELDOUT_ROWS = 100
MAX_TRAIN_LENGTH = 512

SOURCES = {
    "A": {
        "dataset": "b-mc2/sql-create-context",
        "revision": "9d80a6a118b838d9defc3798d659a54a2ac2ff37",
        "file": "sql_create_context_v4.json",
        "license": "cc-by-4.0",
        "gated": False,
        "task": "sql",
        "pool": "train",
        "context_definition": "normalized table schema",
        "category_field": None,
        "mapping": "Instruction: fixed SQL system prompt. Input: CREATE TABLE context and question. Output: SQL answer.",
    },
    "B": {
        "dataset": "google/mobile-actions",
        "revision": "e920309bc2acbc2e99a5e3201cf37df2b9fd9151",
        "file": "dataset.jsonl",
        "license": "cc-by-4.0",
        "gated": False,
        "task": "function",
        "pool": "metadata=train, exactly one tool call",
        "context_definition": "normalized available tool set",
        "category_field": "reference function name",
        "mapping": "Instruction: fixed one-call JSON system prompt with seven tool signatures. Input: date/time, weekday and user utterance. Output: reference function name and argument object.",
    },
    "C": {
        "dataset": "AmazonScience/massive",
        "revision": "ed58ac423a2f4121720918bf5301577edce4ffd3",
        "source_revision": "ff6bd8e4b27c3543e4f8fe2108f32bb95a6f8740",
        "file": "en-US/train/0000.parquet",
        "license": "cc-by-4.0",
        "gated": False,
        "task": "function",
        "pool": "en-US train, at least one annotated slot",
        "context_definition": "normalized published intent",
        "category_field": "published intent",
        "mapping": "Instruction: fixed slot extraction JSON system prompt with allowed slot names. Input: published intent and utterance. Output: extract_slots call with exact annotated text spans grouped by slot name in arrays.",
    },
}

SQL_SYSTEM = (
    "Write one SQL query that answers the question using the supplied table schemas. "
    "Use the supplied table and column names. Return only the SQL query."
)
FUNCTION_SYSTEM = (
    "Select one of the available functions to fulfill the user's request. Use the given date "
    "and weekday for relative times. Return exactly one JSON object with a string key "
    "\"name\" and an object key \"arguments\". Use the function's exact name and argument keys. "
    "Return only JSON. All arguments are strings. Omit optional arguments when no value is given. "
    "Calendar datetime uses YYYY-MM-DDTHH:MM:SS. A map query can be a place, business, or address.\n\n"
    "Available functions (optional arguments have a ?):\n"
)
SLOT_SYSTEM = (
    "Extract the slot values from the request using the supplied intent. "
    'Return exactly one JSON object with "name": "extract_slots" and an "arguments" object. '
    "Each argument key is a slot name from the list below. Its value is an array of verbatim text spans "
    "from the request, in their original order. Use arrays even for a single span. Include only slots "
    "present in the request. Return only JSON.\n\nAllowed slots: "
)

H2_REVIEWS = {
    "A": {
        "source_indices_zero_based": [14592, 3278, 36048, 32098, 29256, 18289, 13434, 71482, 11395, 77397],
        "note": "Ten English questions and SQL answers checked for readability. Row 71482 compares a VARCHAR column to a numeric literal, which may depend on SQL dialect.",
    },
    "B": {
        "source_indices_zero_based": [8137, 912, 204, 2253, 2006, 1828, 1143, 839, 8442, 7366],
        "note": "Ten English requests and reference call names and argument objects checked for readability.",
    },
    "C": {
        "published_row_ids": ["11680", "1992", "543", "13448", "5018", "4427", "4035", "2509", "13276", "1844"],
        "note": "Ten English requests and reference slot annotations checked for readable, verbatim spans.",
    },
}

QUOTES = str.maketrans({character: "'" for character in "\"'`‘’“”"})
SQL_LITERALS = re.compile(r"""'(?:''|[^'])*'|"(?:""|[^"])*" """.strip())
HARD_SQL = re.compile(r"\bJOIN\b|\bGROUP\s+BY\b|\(\s*(?:SELECT|WITH)\b", re.IGNORECASE)
TABLE_NAME = re.compile(r"\bCREATE\s+TABLE\s+(\w+)", re.IGNORECASE)
SLOT_ANNOTATION = re.compile(r"\[([^:\[\]]+?)\s*:\s*([^\[\]]+?)\s*\]")


@dataclass
class Example:
    source_index: int
    system: str
    prompt: str
    answer: str
    context: str | None = None
    category: str | None = None
    copies_prompt_value: bool = False
    utterance: str | None = None
    prompt_ids: list[int] = field(default_factory=list)
    input_ids: list[int] = field(default_factory=list)
    labels: list[int] = field(default_factory=list)

    @property
    def messages(self):
        return [
            {"role": "system", "content": self.system},
            {"role": "user", "content": self.prompt},
            {"role": "assistant", "content": self.answer},
        ]


@dataclass
class PreparedCandidate:
    train: list[Example]
    heldout: list[Example]
    metrics: dict
    system: str
    task: str
    max_new_tokens: int

    def score(self, prediction, answer):
        return score_answer(self.task, prediction, answer)


def normalize_prompt(text):
    return " ".join(text.lower().split())


def normalize_sql(text):
    return normalize_prompt(text.translate(QUOTES)).removesuffix(";").rstrip()


def score_sql(prediction, answer):
    import sqlglot

    try:
        statements = sqlglot.parse(prediction)
        parses = len(statements) == 1 and statements[0] is not None and not isinstance(statements[0], sqlglot.exp.Command)
    except sqlglot.errors.SqlglotError:
        parses = False
    return {"correct": normalize_sql(prediction) == normalize_sql(answer), "parses": parses}


def compact_json(value):
    return json.dumps(value, ensure_ascii=False, separators=(",", ":"), sort_keys=True, allow_nan=False)


def strict_json(value):
    def object_without_duplicates(items):
        result = dict(items)
        if len(result) != len(items):
            raise ValueError("Repeated JSON key")
        return result

    def reject_constant(constant):
        raise ValueError(f"Invalid JSON constant: {constant}")

    return json.loads(value, object_pairs_hook=object_without_duplicates, parse_constant=reject_constant)


def score_function(prediction, answer):
    try:
        decoded = strict_json(prediction)
    except (ValueError, RecursionError):
        return {"correct": False, "json_valid": False, "name_correct": False, "arguments_correct": False}
    gold = json.loads(answer)
    is_object = isinstance(decoded, dict)
    name = is_object and decoded.get("name") == gold["name"]
    arguments = is_object and isinstance(decoded.get("arguments"), dict)
    arguments = arguments and decoded["arguments"] == gold["arguments"]
    return {"correct": name and arguments, "json_valid": True, "name_correct": name, "arguments_correct": arguments}


def score_answer(task, prediction, answer):
    if task == "sql":
        return score_sql(prediction, answer)
    if task == "function":
        return score_function(prediction, answer)
    if task == "classification":
        return {"correct": prediction.strip().lower() == answer.strip().lower()}
    raise ValueError(f"Unsupported scoring task: {task}")


def map_sql(raw, index):
    schema = raw["context"].strip()
    question = raw["question"].strip()
    answer = raw["answer"].strip()
    copies_name = bool(set(TABLE_NAME.findall(schema.lower())) & set(re.findall(r"\w+", answer.lower())))
    return Example(
        index, SQL_SYSTEM, f"Schema:\n{schema}\n\nQuestion:\n{question}", answer,
        context=schema, copies_prompt_value=copies_name,
    )


def render_tool_signatures(tools):
    signatures = []
    for function in tools:
        parameters = function["parameters"]
        required = set(parameters.get("required", []))
        if any(field["type"] != "STRING" for field in parameters["properties"].values()):
            raise ValueError(f"Unexpected non-string field in {function['name']}")
        arguments = [key if key in required else f"{key}?" for key in sorted(parameters["properties"])]
        signatures.append(f"{function['name']}({', '.join(arguments)}): {function['description']}")
    return "\n".join(signatures)


def map_function(raw, index, system, tools):
    developer, user, assistant = raw["messages"]
    timestamp, weekday, _ = developer["content"].splitlines()
    timestamp = timestamp.split(": ", maxsplit=1)[1]
    weekday = weekday.removeprefix("Day of week is ")
    utterance = user["content"].strip()
    function = assistant["tool_calls"][0]["function"]
    if not isinstance(function["arguments"], dict):
        raise ValueError(f"Unexpected non-object arguments at source row {index}")
    copies_value = any(
        isinstance(value, str) and value and value.lower() in utterance.lower()
        for value in function["arguments"].values()
    )
    return Example(
        index, system,
        f"Current date/time: {timestamp}\nDay of week: {weekday}\n\nUser request:\n{utterance}",
        compact_json(function), context=tools, category=function["name"],
        copies_prompt_value=copies_value, utterance=utterance,
    )


def load_slots(table):
    hf_features = json.loads(table.schema.metadata[b"huggingface"])["info"]["features"]
    intents = hf_features["intent"]["names"]
    raw = table.select(["id", "utt", "annot_utt", "intent"]).to_pylist()
    eligible = []
    slot_types = set()
    for index, row in enumerate(raw):
        slots = {}
        for slot_type, span in SLOT_ANNOTATION.findall(row["annot_utt"]):
            slots.setdefault(slot_type, []).append(span)
        if not slots:
            continue
        reconstructed = SLOT_ANNOTATION.sub(r"\2", row["annot_utt"])
        if normalize_prompt(reconstructed) != normalize_prompt(row["utt"]):
            raise ValueError(f"Unexpected slot annotation at published row {row['id']}")
        slot_types.update(slots)
        eligible.append((index, row, slots))

    system = SLOT_SYSTEM + ", ".join(sorted(slot_types))
    rows = []
    for index, row, slots in eligible:
        intent = intents[row["intent"]]
        utterance = row["utt"].strip()
        rows.append(Example(
            index, system, f"Intent: {intent}\nRequest: {utterance}",
            compact_json({"name": "extract_slots", "arguments": slots}),
            context=intent, category=intent, copies_prompt_value=True, utterance=utterance,
        ))
    return raw, rows


def load_candidate(candidate, *, source_file=None):
    """Read the pinned source file, or an identical local copy for offline screening."""
    source = SOURCES[candidate]
    if source_file is None:
        from huggingface_hub import hf_hub_download

        source_file = hf_hub_download(
            repo_id=source["dataset"], repo_type="dataset",
            filename=source["file"], revision=source["revision"],
        )
    if candidate == "A":
        with open(source_file, encoding="utf-8") as file:
            raw = json.load(file)
        rows = [map_sql(row, index) for index, row in enumerate(raw)]
    elif candidate == "B":
        with open(source_file, encoding="utf-8") as file:
            raw = [json.loads(line) for line in file]
        eligible = [
            (index, row) for index, row in enumerate(raw)
            if row["metadata"] == "train" and len(row["messages"][-1]["tool_calls"]) == 1
        ]
        tool_sets = {
            compact_json(sorted((tool["function"] for tool in row["tools"]), key=lambda item: item["name"]))
            for _, row in eligible
        }
        if len(tool_sets) != 1:
            raise ValueError("Expected the same tool set for every eligible Mobile Actions row")
        tools = tool_sets.pop()
        system = FUNCTION_SYSTEM + render_tool_signatures(json.loads(tools))
        rows = [map_function(row, index, system, tools) for index, row in eligible]
    else:
        import pyarrow.parquet as parquet

        raw, rows = load_slots(parquet.read_table(source_file))
    source_revision = source.get("source_revision", source["revision"])
    metadata = {
        **source, "source_rows": len(raw), "excluded_by_source_pool": len(raw) - len(rows),
        "source_url": f"https://huggingface.co/datasets/{source['dataset']}/tree/{source_revision}",
    }
    return rows, metadata


def is_harder_sql(answer):
    return bool(HARD_SQL.search(SQL_LITERALS.sub("", answer)))


def nearest_rank(values, percentile):
    ordered = sorted(values)
    return ordered[math.ceil(len(ordered) * percentile / 100) - 1]


def length_stats(values):
    return {"p50": nearest_rank(values, 50), "p95": nearest_rank(values, 95), "max": max(values)}


def describe_lengths(full_lengths, answer_lengths):
    return {"full": length_stats(full_lengths), "answer": length_stats(answer_lengths)}


def batched_pool_lengths(tokenizer, rows):
    full_lengths, answer_lengths = [], []
    for start in range(0, len(rows), 512):
        batch = rows[start:start + 512]
        rendered = tokenizer.apply_chat_template([row.messages for row in batch], tokenize=False)
        full_ids = tokenizer(rendered, add_special_tokens=False, verbose=False)["input_ids"]
        answer_ids = tokenizer([row.answer for row in batch], add_special_tokens=False, verbose=False)["input_ids"]
        full_lengths.extend(map(len, full_ids))
        answer_lengths.extend(map(len, answer_ids))
    return full_lengths, answer_lengths


def encode_example(tokenizer, row):
    full = tokenizer.apply_chat_template(row.messages, tokenize=False)
    prompt = tokenizer.apply_chat_template(row.messages[:-1], tokenize=False, add_generation_prompt=True)
    row.input_ids = tokenizer.encode(full, add_special_tokens=False)
    row.prompt_ids = tokenizer.encode(prompt, add_special_tokens=False)
    prefix_len = len(row.prompt_ids)
    if row.input_ids[:prefix_len] != row.prompt_ids:
        raise ValueError(f"Template tokenization does not preserve the prompt prefix at row {row.source_index}")
    stop = len(row.input_ids) - row.input_ids[::-1].index(tokenizer.eos_token_id) - 1
    if stop <= prefix_len:
        raise ValueError(f"No answer tokens at row {row.source_index}")
    row.labels = [-100] * prefix_len + row.input_ids[prefix_len:stop + 1] + [-100] * (len(row.input_ids) - stop - 1)


def prepare_candidate(candidate, tokenizer, *, harder=False, loaded=None):
    """Normalize and deduplicate, then sample and prepare train and held-out examples."""
    rows, metadata = loaded if loaded is not None else load_candidate(candidate)
    task = metadata["task"]
    if harder and task != "sql":
        raise ValueError("The harder slice is defined only for SQL")
    scoped = [row for row in rows if is_harder_sql(row.answer)] if harder else rows
    usable = [row for row in scoped if row.prompt.strip() and row.answer.strip()]
    unique = {}
    conflicting_reference_prompts = set()
    for row in usable:
        key = normalize_prompt(row.prompt)
        first = unique.setdefault(key, row)
        if first.answer.lower() != row.answer.lower():
            conflicting_reference_prompts.add(key)
    deduped = list(unique.values())
    if len(deduped) < TRAIN_ROWS + HELDOUT_ROWS:
        raise ValueError(f"{candidate} has too few usable prompts after deduplication: {len(deduped)}")

    full_lengths, answer_lengths = batched_pool_lengths(tokenizer, deduped)
    pool_lengths = describe_lengths(full_lengths, answer_lengths)
    indices = random.Random(SEED).sample(range(len(deduped)), TRAIN_ROWS + HELDOUT_ROWS)
    train = [deduped[index] for index in indices[:TRAIN_ROWS]]
    heldout = [deduped[index] for index in indices[TRAIN_ROWS:]]
    for row in train + heldout:
        encode_example(tokenizer, row)

    train_prompts = {normalize_prompt(row.prompt) for row in train}
    prompt_overlap = sum(normalize_prompt(row.prompt) in train_prompts for row in heldout)
    if prompt_overlap:
        raise ValueError(f"Train/held-out prompt overlap: {prompt_overlap}")
    train_contexts = {normalize_prompt(row.context) for row in train if row.context is not None}
    contexts = [normalize_prompt(row.context) for row in heldout if row.context is not None]
    context_overlap = sum(context in train_contexts for context in contexts) if contexts else None
    n1_fraction = sum(5 <= length <= 100 for length in answer_lengths) / len(answer_lengths)
    copy_fraction = sum(row.copies_prompt_value for row in deduped) / len(deduped)
    near_duplicates = []
    if candidate == "B":
        seen_utterances = {}
        for row in deduped:
            key = (normalize_prompt(row.utterance), row.category)
            first = seen_utterances.setdefault(key, row.source_index)
            if first != row.source_index:
                near_duplicates.append([first, row.source_index])
    review = H2_REVIEWS.get(candidate)
    cap = 2 * pool_lengths["answer"]["p95"]

    metrics = {
        **metadata, "candidate": candidate, "harder_sql_slice": harder,
        "rows_in_slice": len(scoped), "unusable_rows": len(scoped) - len(usable),
        "duplicates_dropped": len(usable) - len(deduped), "usable_dedup_rows": len(deduped),
        "duplicate_prompts_with_conflicting_references": len(conflicting_reference_prompts),
        "near_duplicate_source_index_pairs": near_duplicates,
        "train_rows": len(train), "heldout_rows": len(heldout),
        "heldout_prompt_overlap_count": prompt_overlap,
        "heldout_context_seen_train_count": context_overlap,
        "heldout_context_seen_train_fraction": context_overlap / len(contexts) if contexts else None,
        "pool_lengths": pool_lengths,
        "sampled_lengths": describe_lengths([full_lengths[index] for index in indices], [answer_lengths[index] for index in indices]),
        "train_over_512": sum(len(row.input_ids) > MAX_TRAIN_LENGTH for row in train),
        "heldout_over_512": sum(len(row.input_ids) > MAX_TRAIN_LENGTH for row in heldout),
        "max_new_tokens": cap,
        "h2_review": review,
        "n1_answer_5_to_100_fraction": n1_fraction,
        "n2_prompt_copy_fraction": copy_fraction,
        "hard_requirements": {
            "H1": True, "H2": bool(review) if review is not None else None,
            "H3": True, "H4": len(deduped) >= 2000 and not metadata["gated"],
            "H5": pool_lengths["full"]["p95"] <= MAX_TRAIN_LENGTH, "H6": None, "H7": None,
        },
        "nice_to_have": {
            "N1": n1_fraction >= 0.9, "N2": copy_fraction >= 0.5,
            "N3": metadata["category_field"] is not None, "N4": len(usable) > len(deduped) or bool(near_duplicates),
        },
    }
    return PreparedCandidate(train, heldout, metrics, train[0].system, task, cap)
```

## Greedy evaluation and PEFT fallback

```python
%%writefile /kaggle/temp/soup_bakeoff/probe_runtime.py
"""One-device fp16 inference and the answer-only PEFT probe."""

import math
import random
import time
from collections import Counter

import numpy as np
import torch
import torch.nn.functional as F
from peft import LoraConfig, TaskType, get_peft_model
from transformers import AutoModelForCausalLM, GenerationConfig


SEED = 42
TRAIN_BATCH_SIZE = 8
MAX_TRAIN_LENGTH = 512
GENERATION_BATCH_SIZE = 16


def seed_everything():
    random.seed(SEED)
    np.random.seed(SEED)
    torch.manual_seed(SEED)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(SEED)


def load_model(model_id, revision=None):
    if torch.cuda.device_count() != 1 or "T4" not in torch.cuda.get_device_name(0):
        raise RuntimeError("The probe requires exactly one visible NVIDIA T4.")
    seed_everything()
    return AutoModelForCausalLM.from_pretrained(
        model_id,
        revision=revision,
        dtype=torch.float16,
        attn_implementation="sdpa",
        device_map={"": 0},
    ).eval()


def _left_pad(sequences, pad_id, device):
    width = max(map(len, sequences))
    padding = [width - len(sequence) for sequence in sequences]
    input_ids = [[pad_id] * extra + list(sequence) for extra, sequence in zip(padding, sequences)]
    attention = [[0] * extra + [1] * len(sequence) for extra, sequence in zip(padding, sequences)]
    return (
        torch.tensor(input_ids, device=device),
        torch.tensor(attention, device=device),
    )


def _answer_sequence(row, max_length=None, *, exclude_stop_id=None):
    # No position after the last supervised target can affect the causal loss.
    labels = row.labels[:max_length]
    last_target = next((index for index, label in enumerate(reversed(labels)) if label != -100), None)
    if last_target is None or last_target == len(labels) - 1:
        raise ValueError(f"An example has no predictable answer tokens at max_length={max_length}.")
    end = len(labels) - last_target
    if exclude_stop_id is not None and labels[end - 1] == exclude_stop_id:
        end -= 1
        if not any(label != -100 for label in labels[1:end]):
            raise ValueError("An example has no answer content before its stop token.")
    return row.input_ids[:end], [-100, *labels[1:end]]


def _answer_nll(model, tokenizer, sequences):
    ids, labels = zip(*sequences)
    input_ids, attention = _left_pad(ids, tokenizer.pad_token_id, model.device)
    width = input_ids.shape[1]
    padded_labels = [[-100] * (width - len(row)) + row for row in labels]
    answer_width = max(len(row) - next(i for i, target in enumerate(row) if target != -100) for row in labels)
    targets = torch.tensor([row[-answer_width:] for row in padded_labels], device=model.device)
    # Feeding through the penultimate token aligns the last answer_width logits
    # with the last answer_width labels. Qwen can project only those positions.
    position_ids = (attention.cumsum(dim=1) - 1).clamp_min(0)
    logits = model(
        input_ids=input_ids[:, :-1],
        attention_mask=attention[:, :-1],
        position_ids=position_ids[:, :-1],
        use_cache=False,
        logits_to_keep=answer_width,
    ).logits
    nll = F.cross_entropy(
        logits.flatten(0, 1).float(),
        targets.flatten(),
        reduction="sum",
        ignore_index=-100,
    )
    token_count = sum(target != -100 for row in labels for target in row)
    return nll, token_count


@torch.inference_mode()
def evaluate(model, tokenizer, rows, score, max_new_tokens, log, check_deadline=None):
    started = time.monotonic()
    model.eval()
    eos = tokenizer.eos_token_id
    generation = GenerationConfig(
        do_sample=False,
        repetition_penalty=1.0,
        max_new_tokens=max_new_tokens,
        eos_token_id=eos,
        pad_token_id=tokenizer.pad_token_id,
        use_cache=True,
    )
    judged = [None] * len(rows)
    stopped = 0
    order = sorted(range(len(rows)), key=lambda index: len(rows[index].prompt_ids))
    for offset in range(0, len(order), GENERATION_BATCH_SIZE):
        if check_deadline is not None:
            check_deadline()
        indices = order[offset:offset + GENERATION_BATCH_SIZE]
        input_ids, attention = _left_pad(
            [rows[index].prompt_ids for index in indices], tokenizer.pad_token_id, model.device
        )
        generated = model.generate(input_ids=input_ids, attention_mask=attention, generation_config=generation)
        new_tokens = generated[:, input_ids.shape[1]:].tolist()
        for index, tokens in zip(indices, new_tokens):
            if eos in tokens:
                stop_index = tokens.index(eos)
                stopped += stop_index + 1 < max_new_tokens
                tokens = tokens[:stop_index]
            output = tokenizer.decode(tokens, skip_special_tokens=False, clean_up_tokenization_spaces=False)
            judged[index] = score(output, rows[index].answer)
        log(f"Greedy generation completed {offset + len(indices)}/{len(rows)} held-out rows.")
    generation_seconds = time.monotonic() - started

    total_nll = 0.0
    answer_tokens = 0
    sequences = sorted(
        (_answer_sequence(row, exclude_stop_id=eos) for row in rows),
        key=lambda sequence: len(sequence[0]),
    )
    for offset in range(0, len(sequences), TRAIN_BATCH_SIZE):
        if check_deadline is not None:
            check_deadline()
        nll, count = _answer_nll(model, tokenizer, sequences[offset:offset + TRAIN_BATCH_SIZE])
        total_nll += nll.item()
        answer_tokens += count
    if not math.isfinite(total_nll):
        raise FloatingPointError("Nonfinite teacher-forced answer loss.")

    counts = Counter()
    for judgment in judged:
        counts.update(key for key, value in judgment.items() if value)
    wall_seconds = time.monotonic() - started
    return {
        "score": counts["correct"] / len(rows),
        "stop_rate": stopped / len(rows),
        "answer_loss": total_nll / answer_tokens,
        "answer_tokens": answer_tokens,
        "wall_seconds": wall_seconds,
        "generation_seconds": generation_seconds,
        "teacher_forced_seconds": wall_seconds - generation_seconds,
        "scorer_rates": {
            key: counts[key] / len(rows)
            for key in sorted(judged[0]) if key != "correct"
        },
        "correct": [judgment["correct"] for judgment in judged],
    }


def paired_flips(base, adapted):
    pairs = list(zip(base["correct"], adapted["correct"], strict=True))
    return {
        "wrong_to_right": sum(not before and after for before, after in pairs),
        "right_to_wrong": sum(before and not after for before, after in pairs),
    }


def train_peft(model, tokenizer, rows, log, check_deadline=None):
    started = time.monotonic()
    seed_everything()
    # Match Soup's right cutoff while keeping the sampled training split fixed.
    uncropped = [_answer_sequence(row) for row in rows]
    rendered_overlength = sum(len(row.input_ids) > MAX_TRAIN_LENGTH for row in rows)
    truncated_rows = sum(len(ids) > MAX_TRAIN_LENGTH for ids, _ in uncropped)
    without_targets = sum(
        not any(label != -100 for label in row.labels[1:MAX_TRAIN_LENGTH])
        for row in rows
    )
    if without_targets:
        raise ValueError(
            f"Right truncation to 512 removes all supervised targets from {without_targets} of "
            f"{len(rows)} sampled training rows ({rendered_overlength} rendered rows over 512)."
        )
    cropped = [_answer_sequence(row, MAX_TRAIN_LENGTH) for row in rows]
    supervised_tokens_dropped = sum(
        sum(label != -100 for label in before_labels) - sum(label != -100 for label in after_labels)
        for (_, before_labels), (_, after_labels) in zip(uncropped, cropped, strict=True)
    )
    log(
        f"PEFT training uses {len(rows)} rows. Rendered rows over 512: {rendered_overlength}. "
        f"Rows cropped before the last target: {truncated_rows}. Supervised tokens lost: {supervised_tokens_dropped}."
    )
    adapted = get_peft_model(
        model,
        LoraConfig(
            task_type=TaskType.CAUSAL_LM,
            target_modules="all-linear",
            r=16,
            lora_alpha=32,
            lora_dropout=0.0,
            bias="none",
        ),
    ).train()
    adapted.gradient_checkpointing_enable(gradient_checkpointing_kwargs={"use_reentrant": False})
    log("Enabled non-reentrant gradient checkpointing for the fixed batch size of 8.")
    try:
        params = [param for param in adapted.parameters() if param.requires_grad]
        optimizer = torch.optim.AdamW(params, lr=2e-4, weight_decay=0.0)
        scaler = torch.amp.GradScaler("cuda", init_scale=256.0)
        order = list(range(len(rows)))
        random.Random(SEED).shuffle(order)
        skipped_steps = 0
        optimizer_steps = 0
        for offset in range(0, len(order), TRAIN_BATCH_SIZE):
            indices = order[offset:offset + TRAIN_BATCH_SIZE]
            batch = [cropped[index] for index in indices]
            for _ in range(8):
                if check_deadline is not None:
                    check_deadline()
                optimizer.zero_grad(set_to_none=True)
                nll, tokens = _answer_nll(adapted, tokenizer, batch)
                loss = nll / tokens
                scaler.scale(loss).backward()
                scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(params, max_norm=1.0)
                previous_scale = scaler.get_scale()
                scaler.step(optimizer)
                scaler.update()
                if scaler.get_scale() >= previous_scale:
                    break
                skipped_steps += 1
                log(f"PEFT retried optimizer batch {optimizer_steps + 1} after fp16 gradient overflow.")
            else:
                raise FloatingPointError("Repeated fp16 gradient overflow in the same optimizer batch.")
            optimizer_steps += 1
            if optimizer_steps % 10 == 0 or offset + len(indices) == len(rows):
                log(f"PEFT trained {offset + len(indices)}/{len(rows)} rows. Batch answer loss {loss.item():.4f}.")
        optimizer.zero_grad(set_to_none=True)
        torch.cuda.synchronize()
        adapted.gradient_checkpointing_disable()
        return adapted.eval(), {
            "wall_seconds": time.monotonic() - started,
            "optimizer_steps": optimizer_steps,
            "overflow_retries": skipped_steps,
            "rendered_rows_over_max_length": rendered_overlength,
            "truncated_rows": truncated_rows,
            "supervised_tokens_dropped": supervised_tokens_dropped,
            "trainable_parameters": sum(param.numel() for param in params),
        }
    except Exception:
        adapted.unload()
        raise
```

## Soup probe

Initialize and edit Soup's generated chat config, run its data doctor, then train. The first candidate uses the same fallback as all later candidates if Soup cannot train. [Soup configuration reference](https://trysoup.dev/docs/configuration)

```python
%%writefile /kaggle/temp/soup_bakeoff/soup_probe.py
"""Run the Soup LoRA probe from the generated chat template."""

from collections import deque
from datetime import datetime, timezone
from importlib import metadata
import json
import os
from pathlib import Path
import queue
import re
import shlex
import shutil
import signal
import subprocess
import sys
from threading import Thread
from time import monotonic, time


MODEL = "Qwen/Qwen2.5-1.5B-Instruct"
# Same text as Qwen's shipped template for explicit system/user/assistant
# messages without tool_calls. The generation block defines the loss mask.
MASKED_TEMPLATE = (
    "{% for message in messages %}"
    "{{ '<|im_start|>' + message.role + '\\n' }}"
    "{% if message.role == 'assistant' %}"
    "{% generation %}{{ message.content + '<|im_end|>' }}{% endgeneration %}"
    "{% else %}{{ message.content + '<|im_end|>' }}{% endif %}"
    "{{ '\\n' }}"
    "{% endfor %}"
    "{% if add_generation_prompt %}{{ '<|im_start|>assistant\\n' }}{% endif %}"
)


class SoupProbeError(RuntimeError):
    """Soup did not produce a valid one-epoch adapter."""


class SoupProbeTimeout(SoupProbeError):
    """A Soup stage ran out of time."""


def _run(command, *, cwd, log, deadline, mirror=None, startup_deadline=None):
    command = [str(part) for part in command]
    display = shlex.join(command)
    log(f"Running {display}")
    if monotonic() >= deadline:
        raise SoupProbeTimeout(f"No time left to run {display}")

    env = os.environ.copy()
    env.update(
        PYTHONUNBUFFERED="1",
        PYTHONHASHSEED="42",
        CUDA_VISIBLE_DEVICES="0",
        NO_COLOR="1",
        HF_HUB_DISABLE_TELEMETRY="1",
        SOUP_TELEMETRY="0",
    )
    with subprocess.Popen(
        command,
        cwd=cwd,
        env=env,
        stdin=subprocess.DEVNULL,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        encoding="utf-8",
        errors="replace",
        start_new_session=True,
    ) as process:
        output = queue.Queue()

        def read_output():
            try:
                for line in process.stdout:
                    output.put((monotonic(), line.removesuffix("\n")))
            finally:
                output.put(None)

        reader = Thread(target=read_output, daemon=True)
        reader.start()
        training_seen = False
        trainer_ready = False
        finished = False
        tail = deque(maxlen=12)

        def write_line(line):
            log(line)
            if mirror is not None:
                timestamp = datetime.now(timezone.utc).isoformat(timespec="milliseconds")
                mirror.write(f"{timestamp} {line}\n")
                mirror.flush()
            if line.strip():
                tail.append(line)

        try:
            while True:
                stop_at = deadline
                if startup_deadline is not None and not training_seen:
                    stop_at = min(stop_at, startup_deadline)
                try:
                    if monotonic() >= stop_at and output.empty():
                        raise queue.Empty
                    item = output.get(timeout=max(0, stop_at - monotonic()))
                except queue.Empty:
                    if startup_deadline is not None and not training_seen and stop_at == startup_deadline:
                        raise SoupProbeTimeout(
                            f"Soup did not reach the optimizer loop within the startup budget: {display}"
                        ) from None
                    raise SoupProbeTimeout(f"Deadline expired running {display}") from None
                if item is None:
                    break
                received_at, line = item
                write_line(line)
                if received_at > stop_at:
                    if stop_at == startup_deadline and not training_seen:
                        raise SoupProbeTimeout(
                            f"Soup did not reach the optimizer loop within the startup budget: {display}"
                        )
                    raise SoupProbeTimeout(f"Deadline expired running {display}")
                if "Training started!" in line:
                    trainer_ready = True
                if (
                    startup_deadline is not None
                    and not training_seen
                    and trainer_ready
                    and re.search(r"\b\d+/63\s*\[", line)
                ):
                    # tqdm first appears after HF Trainer enters the loop.
                    training_seen = True
                    log("Soup entered the optimizer loop (63 steps expected).")
                if "Training Complete!" in line:
                    finished = True

            status = process.wait(timeout=max(0, deadline - monotonic()))
            if status:
                recent_output = "\n".join(tail)
                raise SoupProbeError(f"{display} exited {status}. Last output:\n{recent_output}")
            return training_seen, finished
        finally:
            if process.poll() is None:
                try:
                    os.killpg(process.pid, signal.SIGKILL)
                except ProcessLookupError:
                    pass
                process.wait()
            # Keep late process output, including lines written before a timeout.
            reader.join(timeout=2)
            while not output.empty():
                item = output.get_nowait()
                if item is not None:
                    write_line(item[1])


def setup_soup(output_dir, log, *, timeout_seconds):
    """Install Soup into an isolated env reusing Kaggle's installed torch."""
    deadline = monotonic() + timeout_seconds
    output_dir = Path(output_dir).resolve()
    output_dir.mkdir(parents=True, exist_ok=True)
    env_dir = output_dir / ".soup_env"
    # Keep the environment out of /kaggle/working and the notebook output archive.
    if Path("/kaggle/working").is_dir():
        env_dir = Path("/tmp/dataset_bakeoff_soup_env")
    uv = shutil.which("uv")
    create_environment = (
        [uv, "venv", "--seed", "--system-site-packages", "--python", sys.executable, env_dir]
        if uv else [sys.executable, "-m", "venv", "--system-site-packages", env_dir]
    )
    _run(
        create_environment,
        cwd=output_dir,
        log=log,
        deadline=deadline,
    )
    python = env_dir / "bin" / "python"
    soup = env_dir / "bin" / "soup"
    constraints = output_dir / "soup_constraints.txt"
    # Reuse the existing CUDA build or fail fast on Soup's incompatible torch floor.
    constraints.write_text(f"torch=={metadata.version('torch')}\n", encoding="utf-8")
    _run(
        [python, "-m", "pip", "install", "-c", constraints, "soup-cli[train]==0.77.2"],
        cwd=output_dir,
        log=log,
        deadline=deadline,
    )
    _run([soup, "--version"], cwd=output_dir, log=log, deadline=deadline)
    _run(
        [python, "-c", (
            "import sys, torch, transformers, peft, trl;"
            "print(sys.version);"
            "print('torch', torch.__version__, 'transformers', transformers.__version__,"
            "'peft', peft.__version__, 'trl', trl.__version__)"
        )],
        cwd=output_dir,
        log=log,
        deadline=deadline,
    )
    return soup


def _verify_mask(prepared, tokenizer, log):
    """Check Qwen rendering and the full answer/EOT mask before Soup launches."""
    rendered_overlength = 0
    truncated_rows = 0
    supervised_tokens_dropped = 0
    for index, example in enumerate(prepared.train, start=1):
        messages = list(example.messages)
        original = tokenizer.apply_chat_template(messages, tokenize=False)
        masked = tokenizer.apply_chat_template(
            messages, chat_template=MASKED_TEMPLATE, tokenize=False
        )
        if masked != original:
            raise SoupProbeError(f"Row {index}: masked Qwen template changed the rendered text.")
        encoded = tokenizer.apply_chat_template(
            messages,
            chat_template=MASKED_TEMPLATE,
            tokenize=True,
            return_dict=True,
            return_assistant_tokens_mask=True,
        )
        target_ids = [
            token for token, keep in zip(encoded["input_ids"], encoded["assistant_masks"])
            if keep
        ]
        if tokenizer.decode(
            target_ids, skip_special_tokens=False, clean_up_tokenization_spaces=False
        ) != messages[-1]["content"] + "<|im_end|>":
            raise SoupProbeError(f"Row {index}: Soup's answer/EOT loss mask did not match.")
        labels = [
            token if keep else -100
            for token, keep in zip(encoded["input_ids"], encoded["assistant_masks"])
        ]
        if hasattr(example, "labels") and list(example.labels) != labels:
            raise SoupProbeError(f"Row {index}: Soup's mask differs from the evaluation loss mask.")
        if not any(target != -100 for target in labels[1:512]):
            raise SoupProbeError(
                f"Row {index}: Soup's right truncation at 512 tokens would remove all answer/EOT targets."
            )
        rendered_overlength += len(labels) > 512
        dropped = sum(target != -100 for target in labels[512:])
        truncated_rows += dropped > 0
        supervised_tokens_dropped += dropped
    log(f"Verified equivalent Qwen rendering and answer/EOT masks for {len(prepared.train)} rows.")
    log(
        f"Soup training uses {len(prepared.train)} rows. Rendered rows over 512: "
        f"{rendered_overlength}. Rows cropped before the last target: {truncated_rows}. "
        f"Supervised tokens lost: {supervised_tokens_dropped}."
    )
    return {
        "rendered_rows_over_max_length": rendered_overlength,
        "truncated_rows": truncated_rows,
        "supervised_tokens_dropped": supervised_tokens_dropped,
    }


def train_soup(
    prepared, output_dir, log, *, soup, tokenizer, timeout_seconds, startup_timeout_seconds=None,
    expected_revision=None,
):
    """Train one candidate and return the directory containing its PEFT adapter."""
    import yaml

    start = monotonic()
    deadline = start + timeout_seconds
    startup_deadline = None if startup_timeout_seconds is None else start + startup_timeout_seconds
    setup_deadline = deadline if startup_deadline is None else min(deadline, startup_deadline)
    output_dir = Path(output_dir).resolve()
    output_dir.mkdir(parents=True, exist_ok=True)
    train_path = output_dir / "soup_train.jsonl"
    with train_path.open("w", encoding="utf-8") as handle:
        for example in prepared.train:
            handle.write(json.dumps({"messages": list(example.messages)}, ensure_ascii=False) + "\n")

    _run(
        [soup, "init", "--template", "chat"],
        cwd=output_dir,
        log=log,
        deadline=setup_deadline,
    )
    config_path = output_dir / "soup.yaml"
    generated = config_path.read_text(encoding="utf-8")
    (output_dir / "soup_template.yaml").write_text(generated, encoding="utf-8")
    config = yaml.safe_load(generated)
    config["base"] = MODEL
    config["backend"] = "transformers"
    config["data"].update(
        train=str(train_path),
        format="chatml",
        val_split=0.0,
        max_length=512,
        train_on_responses_only=True,
        chat_template=MASKED_TEMPLATE,
    )
    config["training"].update(
        epochs=1,
        lr=2e-4,
        batch_size=8,
        gradient_accumulation_steps=1,
        quantization="none",
        train_on_eot=True,
        seed=42,
        data_seed=42,
        optimizer="adamw_torch",
        scheduler="constant",
        warmup_ratio=0.0,
        weight_decay=0.0,
        max_grad_norm=1.0,
        rewind_log=False,
    )
    config["training"]["lora"].update(r=16, alpha=32, target_modules="all-linear", dropout=0.0)
    adapter = output_dir / "soup_adapter"
    config["output"] = str(adapter)
    config_path.write_text(yaml.safe_dump(config, sort_keys=False), encoding="utf-8")

    with (output_dir / "soup_data_doctor.log").open("w", encoding="utf-8") as doctor_log:
        try:
            _run(
                [
                    soup, "data", "doctor", train_path, "--model", MODEL,
                    "--max-length", "512", "--train-on-eot",
                ],
                cwd=output_dir,
                log=log,
                deadline=setup_deadline,
                mirror=doctor_log,
            )
        except SoupProbeTimeout:
            raise
        except SoupProbeError as error:
            log(f"Soup data doctor returned a failure: {error}. Checking the configured loss mask.")
    truncation = _verify_mask(prepared, tokenizer, log)
    if monotonic() >= setup_deadline:
        raise SoupProbeTimeout("Soup startup deadline expired before launching soup train --yes.")

    start_epoch = time()
    command_start = monotonic()
    try:
        training_seen, finished = _run(
            [soup, "train", "--yes"],
            cwd=output_dir,
            log=log,
            deadline=deadline,
            startup_deadline=startup_deadline,
        )
    finally:
        window = {
            "start_epoch": start_epoch,
            "end_epoch": time(),
            "wall_seconds": monotonic() - command_start,
            **truncation,
        }
        (output_dir / "soup_training_window.json").write_text(
            json.dumps(window, indent=2) + "\n", encoding="utf-8"
        )
    if not finished:
        raise SoupProbeError("Soup exited without confirming the training epoch completed.")
    if startup_deadline is not None and not training_seen:
        raise SoupProbeError("Soup did not show optimizer-loop startup within the startup budget.")
    if not (adapter / "adapter_config.json").is_file() or not any(
        (adapter / filename).is_file() for filename in ("adapter_model.safetensors", "adapter_model.bin")
    ):
        raise SoupProbeError(f"Soup did not save PEFT adapter weights at {adapter}.")
    if expected_revision is not None:
        provenance_path = adapter / "soup_run.json"
        if not provenance_path.is_file():
            raise SoupProbeError("Soup did not emit the base-model revision in soup_run.json.")
        provenance = json.loads(provenance_path.read_text(encoding="utf-8"))
        actual = provenance.get("base_revision")
        if actual != expected_revision:
            raise SoupProbeError(
                f"Soup trained on base revision {actual!r}, expected {expected_revision!r}."
            )
        log(f"Soup used the same pinned Qwen revision as evaluation: {actual}.")
    return adapter
```

```python
%%writefile /kaggle/temp/soup_bakeoff/probe_report.py
"""Write the Markdown and JSON views of the same evidence."""
import json
from pathlib import Path


HARD = {
    "H1": "automatic scorer",
    "H2": "English and manually checkable",
    "H3": "instruction mapping",
    "H4": "2,000 deduplicated rows, open license, ungated",
    "H5": "rendered p95 ≤512 tokens",
    "H6": "base score 20–70%",
    "H7": "gain ≥10 percentage points",
}


def conclude(result):
    accepted = [
        (row["harder_slice"] if row.get("decision_uses_harder_slice") else row)
        for row in result["candidates"]
    ]
    accepted = [
        row for row in accepted
        if all(row.get("criteria", {}).get(key) is True for key in HARD)
    ]
    if accepted:
        selected = max(accepted, key=lambda row: (
            round(row["gain_pp"], 8),
            sum(value is True for key, value in row["criteria"].items() if key.startswith("N")),
            -row["data"]["pool_lengths"]["full"]["p95"],
            -row["data"]["pool_lengths"]["full"]["p50"],
        ))
        scope = f" on the {selected['slice']} slice" if selected.get("slice") else ""
        result["recommendation"] = f"Pick `{selected['dataset']}`{scope} ({selected['gain_pp']:+.0f} pp on 100 held-out examples)."
        result["recommended_candidate"] = selected["id"]
    else:
        def hard_failure(row):
            return any(row.get("criteria", {}).get(key) is False for key in HARD)

        rejected = all(
            hard_failure(row) and (
                not row.get("harder_slice") or hard_failure(row["harder_slice"])
            )
            for row in result["candidates"]
        )
        result["recommendation"] = (
            "No dataset passed all hard criteria." if rejected else
            "No recommendation yet because the required evidence is incomplete."
        )
        result["recommended_candidate"] = None


def _pct(value):
    return "—" if value is None else f"{value:.1%}"


def _number(value, places=0):
    return "—" if value is None else f"{value:,.{places}f}"


def _triplet(stats):
    return " / ".join(_number(stats.get(key)) for key in ("p50", "p95", "max"))


def _evaluation(evaluation):
    if not evaluation:
        return "—"
    fields = [
        f"score {_pct(evaluation.get('score'))}",
        f"stop {_pct(evaluation.get('stop_rate'))}",
        f"loss {_number(evaluation.get('answer_loss'), 4)}",
    ]
    fields += [f"{key} {_pct(value)}" for key, value in evaluation.get("scorer_rates", {}).items()]
    fields += [f"{_number(evaluation.get('wall_seconds'), 1)} s"]
    return "<br>".join(fields)


def _cell(value):
    return str(value).replace("|", "&#124;").replace("\n", " ")


def _candidate_cells(row):
    data = row.get("data", {})
    training = row.get("training", {})
    flips = row.get("flips", {})
    criteria = row.get("criteria", {})
    statuses = {None: "—", True: "Pass", False: "Fail"}
    label = f"{row['id']} · `{row['dataset']}`"
    if row.get("slice"):
        label += f"<br>{row['slice']}"
    return [
        label,
        row.get("state", "Pending"),
        data.get("license") or "—",
        _number(data.get("source_rows")),
        _number(data.get("excluded_by_source_pool")),
        _number(data.get("rows_in_slice")),
        _number(data.get("unusable_rows")),
        _number(data.get("duplicates_dropped")),
        _number(data.get("duplicate_prompts_with_conflicting_references")),
        _number(data.get("usable_dedup_rows")),
        _number(data.get("train_rows")),
        _number(data.get("heldout_rows")),
        _number(data.get("heldout_prompt_overlap_count")),
        _pct(data.get("heldout_context_seen_train_fraction")),
        _triplet(data.get("pool_lengths", {}).get("full", {})),
        _triplet(data.get("pool_lengths", {}).get("answer", {})),
        _triplet(data.get("sampled_lengths", {}).get("full", {})),
        _triplet(data.get("sampled_lengths", {}).get("answer", {})),
        f"{_number(data.get('train_over_512'))} / {_number(data.get('heldout_over_512'))}",
        _number(data.get("max_new_tokens")),
        _pct(data.get("n1_answer_5_to_100_fraction")),
        _pct(data.get("n2_prompt_copy_fraction")),
        _evaluation(row.get("base")),
        _evaluation(row.get("adapted")),
        _number(row.get("gain_pp"), 1),
        _number(flips.get("wrong_to_right")),
        _number(flips.get("right_to_wrong")),
        training.get("backend", "—"),
        _number(training.get("wall_seconds"), 1),
        _number(training.get("peak_gpu_mib")),
        *[statuses[criteria.get(key)] for key in HARD],
        *[("—" if criteria.get(key) is None else int(criteria[key])) for key in ("N1", "N2", "N3", "N4")],
    ]


def write_results(result, output_dir):
    output_dir = Path(output_dir)
    output_dir.mkdir(parents=True, exist_ok=True)
    conclude(result)
    headings = [
        "Candidate", "State", "License", "Source rows", "Source exclusions", "In slice",
        "Unusable", "Dedup drops", "Conflicting duplicate prompts", "Usable",
        "Train", "Held out", "Prompt overlap", "Context overlap",
        "Pool full tokens", "Pool answer tokens", "Sample full tokens", "Sample answer tokens",
        "Over 512 train / held out", "Generation cap", "Answers 5–100 tokens", "Copy a prompt value",
        "Base evaluation", "LoRA evaluation",
        "Gain (pp)", "Wrong → right", "Right → wrong", "Trainer", "Train (s)", "Peak GPU (MiB)",
        *HARD, "N1", "N2", "N3", "N4",
    ]
    rows = []
    for row in result["candidates"]:
        rows.append(_candidate_cells(row))
        if row.get("harder_slice"):
            rows.append(_candidate_cells(row["harder_slice"]))
    table = [
        "| " + " | ".join(headings) + " |",
        "| " + " | ".join(["---"] * len(headings)) + " |",
        *["| " + " | ".join(map(_cell, row)) + " |" for row in rows],
    ]
    text = [
        "# Dataset bake-off",
        "",
        result["recommendation"],
        "",
        *table,
        "",
        "Token columns show p50 / p95 / max. Rates use the same 100 held-out rows for base and LoRA. "
        "Loss is mean answer-token NLL. A dash indicates an unavailable or inapplicable value.",
        "",
        "## Issues and deviations",
        "",
    ]
    issues = list(result.get("issues", []))
    issues.extend(f"{item['dataset']} — {item['reason']}" for item in result.get("screened_out", []))
    for row in result["candidates"]:
        for part in [row, row.get("harder_slice", {})]:
            issues.extend(f"{part['id']} ({part.get('slice', 'main')}) — {issue}" for issue in part.get("issues", []))
    text += [f"- {issue}" for issue in issues] or ["- None recorded."]
    text += ["", "## Sources", ""]
    for row in result["candidates"]:
        source = row.get("data", {}).get("source_url")
        url = source or f"https://huggingface.co/datasets/{row['dataset']}"
        text.append(f"- {row['id']} [{row['dataset']}]({url})")
    for name, body in {
        "bakeoff_results.json": json.dumps(result, ensure_ascii=False, allow_nan=False, indent=2) + "\n",
        "bakeoff_results.md": "\n".join(text) + "\n",
    }.items():
        target = output_dir / name
        pending = target.with_suffix(target.suffix + ".tmp")
        pending.write_text(body, encoding="utf-8")
        pending.replace(target)
```

```python
%%writefile /kaggle/temp/soup_bakeoff/run_probe.py
"""Run one controlled bake-off on a Kaggle T4."""
import gc
import importlib.metadata
import json
import os
from pathlib import Path
import platform
import signal
import subprocess
import sys
import time
import traceback

from probe_report import write_results


MODEL = "Qwen/Qwen2.5-1.5B-Instruct"
MODEL_REVISION = "989aa7980e4cf806f80c7fef2b1adb7bc71aa306"
OUTPUT = Path("/kaggle/working")
SCRATCH = Path("/kaggle/temp/soup_bakeoff")
CONFIG = SCRATCH / "probe_config.json"


class BudgetExpired(BaseException):
    pass


def log(message):
    # The notebook wrapper timestamps every line including third-party stdout/stderr.
    print(message, flush=True)


def dependency_versions():
    for package in ("torch", "transformers", "peft", "datasets", "sqlglot"):
        try:
            version = importlib.metadata.version(package)
        except importlib.metadata.PackageNotFoundError:
            version = "missing"
        log(f"{package} {version}")


def stop_at_budget(*_):
    raise BudgetExpired("The bake-off reached its time limit.")


def peak_gpu_memory(samples, start, end):
    from datetime import datetime

    values = []
    for line in samples.read_text(encoding="utf-8").splitlines()[1:]:
        try:
            date, _, _, memory = line.split(",", 3)
            timestamp = datetime.strptime(date.strip(), "%Y/%m/%d %H:%M:%S.%f").timestamp()
            if start <= timestamp <= end:
                values.append(int(memory.split()[0]))
        except (ValueError, IndexError):
            continue
    return max(values, default=None)


def phase(row, description, result):
    row["state"] = description
    log(f"{row['id']} {row.get('slice', 'main')} | {description}")
    write_results(result, OUTPUT)


def log_preparation(prepared, row, tokenizer):
    row["data"].update(prepared.metrics)
    row["criteria"].update(prepared.metrics["hard_requirements"])
    row["criteria"].update(prepared.metrics["nice_to_have"])
    log(json.dumps(prepared.metrics, ensure_ascii=False))
    first_train = prepared.train[0]
    first_eval = prepared.heldout[0]
    log("Fully rendered training example")
    log(tokenizer.apply_chat_template(first_train.messages, tokenize=False))
    log("Rendered evaluation prompt")
    log(tokenizer.apply_chat_template(first_eval.messages[:-1], tokenize=False, add_generation_prompt=True))


def run(result, candidates, deadline, samples, backend):
    import torch
    from peft import PeftModel
    from transformers import AutoTokenizer, set_seed
    from probe_data import prepare_candidate
    from probe_runtime import evaluate, load_model, paired_flips, train_peft
    from soup_probe import setup_soup, train_soup

    set_seed(42)
    if torch.cuda.device_count() != 1 or "T4" not in torch.cuda.get_device_name(0):
        raise RuntimeError("The probe requires exactly one visible NVIDIA T4 GPU.")
    tokenizer = AutoTokenizer.from_pretrained(MODEL, revision=MODEL_REVISION)
    soup_environment = None

    def release_gpu():
        gc.collect()
        torch.cuda.empty_cache()

    def test(prepared, row):
        nonlocal backend, soup_environment
        phase(row, "Base evaluation", result)
        # Each probe begins with the same pinned, untouched base weights.
        model = load_model(MODEL, MODEL_REVISION)
        base = evaluate(model, tokenizer, prepared.heldout, prepared.score, prepared.max_new_tokens, log)
        row["base"] = base
        row["criteria"]["H6"] = 0.2 <= base["score"] <= 0.7
        log(json.dumps({key: value for key, value in base.items() if key != "correct"}))
        phase(row, "LoRA probe", result)
        if backend == "Soup":
            del model
            release_gpu()
            start = time.time()
            first_attempt = soup_environment is None
            startup_deadline = min(start + 900, deadline)
            soup_directory = SCRATCH / f"soup_{row['id']}_{'harder' if row.get('slice') else 'main'}"
            try:
                if first_attempt:
                    soup_environment = setup_soup(
                        SCRATCH / "soup_environment", log,
                        timeout_seconds=max(1, startup_deadline - time.time()),
                    )
                adapter = train_soup(
                    prepared, soup_directory,
                    log, tokenizer=tokenizer, soup=soup_environment,
                    expected_revision=MODEL_REVISION,
                    startup_timeout_seconds=max(1, startup_deadline - time.time()) if first_attempt else None,
                    timeout_seconds=max(1, deadline - time.time()),
                )
                window = json.loads((soup_directory / "soup_training_window.json").read_text(encoding="utf-8"))
                row["training"] = {
                    **window,
                    "backend": backend,
                    "peak_gpu_mib": peak_gpu_memory(samples, window["start_epoch"], window["end_epoch"]),
                }
                model = PeftModel.from_pretrained(load_model(MODEL, MODEL_REVISION), adapter)
            except Exception as exc:
                log(traceback.format_exc())
                if not first_attempt:
                    raise
                backend = "Transformers + PEFT"
                note = f"First Soup attempt failed after {time.time() - start:.1f} s: {type(exc).__name__}: {exc}. Switched all probes to Transformers + PEFT."
                result["issues"].append(note)
                log(note)
                model = load_model(MODEL, MODEL_REVISION)
        if backend != "Soup":
            set_seed(42)
            start = time.time()
            model, metrics = train_peft(model, tokenizer, prepared.train, log)
            finish = time.time()
            row["training"] = {
                **metrics,
                "backend": backend,
                "wall_seconds": finish - start,
                "peak_gpu_mib": peak_gpu_memory(samples, start, finish),
            }
        log(json.dumps(row["training"]))
        phase(row, "Adapted evaluation", result)
        adapted = evaluate(model, tokenizer, prepared.heldout, prepared.score, prepared.max_new_tokens, log)
        row["adapted"] = adapted
        row["gain_pp"] = round(100 * (adapted["score"] - base["score"]), 8)
        row["criteria"]["H7"] = row["gain_pp"] >= 10
        row["flips"] = paired_flips(base, adapted)
        log(json.dumps({key: value for key, value in adapted.items() if key != "correct"}))
        log(json.dumps(row["flips"]))
        del model
        release_gpu()
        phase(row, "Complete", result)

    for row in candidates:
        try:
            phase(row, "Preparing dataset", result)
            prepared = prepare_candidate(row["id"], tokenizer)
            log_preparation(prepared, row, tokenizer)
            if any(row["criteria"].get(key) is False for key in ("H1", "H2", "H3", "H4", "H5")):
                phase(row, "Rejected during screening", result)
                continue
            test(prepared, row)
            if row["id"] == "A" and row["base"]["score"] > 0.7:
                harder = {
                    "id": row["id"], "dataset": row["dataset"], "slice": "JOIN, GROUP BY or subquery",
                    "criteria": {key: value for key, value in row["criteria"].items() if key not in ("H5", "H6", "H7")},
                    "data": {"source_url": row["data"].get("source_url"), "license": row["data"].get("license")},
                }
                row["harder_slice"] = harder
                phase(harder, "Preparing dataset", result)
                prepared = prepare_candidate(row["id"], tokenizer, harder=True)
                log_preparation(prepared, harder, tokenizer)
                test(prepared, harder)
                if all(harder["criteria"].get(f"H{n}") is True for n in range(1, 8)):
                    row["decision_uses_harder_slice"] = True
                    write_results(result, OUTPUT)
        except Exception as exc:
            failed = row.get("harder_slice") or row
            failed.setdefault("issues", []).append(f"{type(exc).__name__}: {exc}")
            log(traceback.format_exc())
            release_gpu()
            phase(failed, "Failed", result)


def write_agent_log(result):
    lines = [
        "# Agent log", "",
        "- Used seed 42 and a single visible T4 with the pinned Qwen revision.",
        "- Used nearest-rank percentiles across the deduplicated pool and reported the 600-row sample separately.",
        "- Set the generation cap at twice the pool p95 for answer tokens.",
        "- Supervised the assistant end-of-turn token during training. Reported held-out loss on answer content tokens and measured stopping separately.",
        "- Used non-reentrant gradient checkpointing in the PEFT probe to keep the same batch size of 8 on one T4.",
        "- Marked N1 when at least 90% of answers were between 5 and 100 tokens.",
        "- Used p95 rendered length, then p50, to resolve a remaining sequence-length tie.",
        "- Reported normalized prompt overlap and reused context overlap independently.",
        "- Reported the sampled device memory during the training window.",
        "- Used the signed-in Kaggle website after the connector returned Unauthenticated. The T4 selector showed 30 GPU hours remaining before launch.",
        "",
        "## Issues and deviations", "",
        *[f"- {issue}" for issue in result.get("issues", [])],
    ]
    for row in result["candidates"]:
        for part in [row, row.get("harder_slice", {})]:
            lines.extend(f"- {part['id']} ({part.get('slice', 'main')}) — {issue}" for issue in part.get("issues", []))
    (OUTPUT / "agent_log.md").write_text("\n".join(lines) + "\n", encoding="utf-8")


def main():
    config = json.loads(CONFIG.read_text(encoding="utf-8"))
    result = {
        "model": MODEL, "model_revision": MODEL_REVISION, "seed": 42,
        "environment": "Kaggle T4", "candidates": config["candidates"],
        "issues": config.get("issues", []), "screened_out": config.get("screened_out", []),
    }
    deadline = config["deadline_epoch"]
    log(f"Runtime deadline (UTC epoch seconds) {deadline}")
    log(f"Python {platform.python_version()}")
    dependency_versions()
    samples = OUTPUT / "nvidia_smi_samples.csv"
    sampler = None
    try:
        signal.signal(signal.SIGALRM, stop_at_budget)
        signal.setitimer(signal.ITIMER_REAL, max(0.001, deadline - time.time()))
        with samples.open("w", encoding="utf-8") as handle:
            sampler = subprocess.Popen(
                ["nvidia-smi", "-i", "0", "--query-gpu=timestamp,index,name,memory.used", "--format=csv", "-lms", "1000"],
                stdout=handle,
            )
            run(result, result["candidates"], deadline, samples, config.get("initial_backend", "Soup"))
    except BudgetExpired as exc:
        result["issues"].append(str(exc))
        log(str(exc))
    except Exception as exc:
        result["issues"].append(f"{type(exc).__name__}: {exc}")
        log(traceback.format_exc())
    finally:
        signal.setitimer(signal.ITIMER_REAL, 0)
        if sampler:
            sampler.terminate()
            sampler.wait(timeout=5)
        write_results(result, OUTPUT)
        write_agent_log(result)


if __name__ == "__main__":
    main()
```

```python
%%writefile /kaggle/temp/soup_bakeoff/notebook_entry.py
"""Set up a dedicated probe environment and retain its complete output."""
from datetime import datetime, timezone
import importlib.metadata
import json
import os
from pathlib import Path
import shutil
import signal
import subprocess
import sys
import threading
import time
import traceback

from probe_report import write_results
from run_probe import CONFIG, MODEL, MODEL_REVISION, OUTPUT, SCRATCH, write_agent_log


def run_logged(command, emit, deadline):
    emit("$ " + " ".join(map(str, command)))
    remaining = deadline - time.time()
    if remaining <= 0:
        raise TimeoutError("The bake-off reached its time limit before this command.")
    with subprocess.Popen(
        command, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
        text=True, encoding="utf-8", errors="replace", start_new_session=True,
    ) as process:
        def stop():
            try:
                os.killpg(process.pid, signal.SIGKILL)
            except ProcessLookupError:
                pass

        watchdog = threading.Timer(remaining, stop)
        watchdog.start()
        try:
            for line in process.stdout:
                emit(line.removesuffix("\n"))
            status = process.wait()
        finally:
            watchdog.cancel()
            if process.poll() is None:
                stop()
                process.wait()
        if status:
            raise subprocess.CalledProcessError(status, command)


def main():
    OUTPUT.mkdir(parents=True, exist_ok=True)
    config = json.loads(CONFIG.read_text(encoding="utf-8"))
    deadline = min(time.time() + config["runtime_seconds"], config.get("deadline_epoch", float("inf")))
    # Reserve a brief window to persist partial metrics after the runner's alarm.
    config["deadline_epoch"] = deadline - 15
    CONFIG.write_text(json.dumps(config), encoding="utf-8")
    os.environ.update(
        CUDA_VISIBLE_DEVICES="0",
        PYTHONHASHSEED="42",
        PYTHONUNBUFFERED="1",
        TOKENIZERS_PARALLELISM="false",
        HF_HUB_DISABLE_TELEMETRY="1",
        WANDB_MODE="disabled",
    )
    result = {
        "model": MODEL, "model_revision": MODEL_REVISION, "seed": 42,
        "environment": "Kaggle T4", "candidates": config["candidates"],
        "issues": config.get("issues", []), "screened_out": config.get("screened_out", []),
    }
    with (OUTPUT / "bakeoff.log").open("a", encoding="utf-8", buffering=1) as log:
        def emit(message):
            for line in message.split("\n"):
                timestamp = datetime.now(timezone.utc).isoformat(timespec="milliseconds")
                entry = f"{timestamp} {line}"
                log.write(entry + "\n")
                print(entry, flush=True)

        try:
            run_logged(["nvidia-smi"], emit, deadline)
            run_logged([sys.executable, "--version"], emit, deadline)
            requirements = SCRATCH / "constraints.txt"
            requirements.write_text(
                f"torch=={importlib.metadata.version('torch')}\n", encoding="utf-8"
            )
            environment = SCRATCH / "probe-venv"
            uv = shutil.which("uv")
            if uv:
                run_logged([
                    uv, "venv", "--seed", "--system-site-packages", "--python", sys.executable, environment,
                ], emit, deadline)
            else:
                run_logged([sys.executable, "-m", "venv", "--system-site-packages", environment], emit, deadline)
            python = environment / "bin/python"
            run_logged([
                python, "-m", "pip", "--disable-pip-version-check", "install", "-c", requirements,
                "transformers==4.57.1", "peft==0.18.0", "accelerate==1.11.0", "sqlglot==28.0.0",
            ], emit, deadline)
            run_logged([python, SCRATCH / "run_probe.py"], emit, deadline)
        except (Exception, KeyboardInterrupt) as exc:
            emit(traceback.format_exc())
            detail = f"{type(exc).__name__}: {exc}. See bakeoff.log."
            existing = OUTPUT / "bakeoff_results.json"
            if existing.exists():
                result = json.loads(existing.read_text(encoding="utf-8"))
            result["issues"].append(detail)
            write_results(result, OUTPUT)
            write_agent_log(result)


if __name__ == "__main__":
    main()
```
